```python
df["Distance_KM"] = df["Distance_KM"].fillna(
    df["Distance_KM"].median()
)
```

```python
df["Payment_Mode"] = df["Payment_Mode"].fillna(
    df["Payment_Mode"].mode()[0]
)
```

```python
df["Payment_Mode"]
```

```python
df["Profit_INR"] = (
    df["Revenue_INR"] - df["Operating_Cost_INR"]
)
```

```python
df["Profit_INR"]
```

```python
df["Fuel_Efficiency"] = (
    df["Distance_KM"] / df["Fuel_Consumed_Liters"]
)
```

```python
df["Fuel_Efficiency"]
```

```python
df["Month"] = pd.to_datetime(df["Trip_Date"]).dt.month_name()
```

```python
df["Day"] = pd.to_datetime(df["Trip_Date"]).dt.day_name()
```

```python
import seaborn as sns

sns.scatterplot(
    data=df,
    x="Distance_KM",
    y="Revenue_INR",
    hue="Vehicle_Type"
)
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x='Vehicle_Type'
)

plt.title('Trip Count by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Number of Trips')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Fuel_Type'
)

plt.title('Number of Trips by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Trips')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Vehicle_Type',
    y='Revenue_INR'
)

plt.title('Revenue Distribution by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Revenue (INR)')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x='Vehicle_Type',
    y='Customer_Rating'
)

plt.title('Customer Rating Distribution by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Customer Rating')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Distance_KM',
    y='Revenue_INR',
    hue='Vehicle_Type'
)

plt.title('Revenue vs Distance')
plt.xlabel('Distance (KM)')
plt.ylabel('Revenue (INR)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Profit_INR',
    bins=40,
    kde=True
)

plt.title('Profit Distribution')
plt.xlabel('Profit (INR)')
plt.ylabel('Number of Trips')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Distance_KM',
    y='Operating_Cost_INR',
    hue='Vehicle_Type'
)

plt.title('Distance vs Operating Cost')
plt.xlabel('Distance (KM)')
plt.ylabel('Operating Cost (INR)')
plt.show()
```
